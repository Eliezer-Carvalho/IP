```python
from transformers import Wav2Vec2ForCTC, Wav2Vec2Processor

from transformers.audio_utils import load_audio

import torch
```

<a href = "https://huggingface.co/jonatasgrosman/wav2vec2-large-xlsr-53-portuguese"> <h1> wav2vec2-large-xlsr-53-portuguese </h1> </a>

```python
MODELO_NAME = "jonatasgrosman/wav2vec2-large-xlsr-53-portuguese"

PROCESSADOR = Wav2Vec2Processor.from_pretrained (MODELO_NAME)
MODELO = Wav2Vec2ForCTC.from_pretrained (MODELO_NAME)

print (PROCESSADOR)
print (MODELO)
```

<HR>

```python
audio = load_audio (r"C:\Users\Admin\Desktop\whisper\audio_2.wav", sampling_rate = PROCESSADOR.feature_extractor.sampling_rate)

print (audio)
```

<hr>

```python
inputs = PROCESSADOR (audio, sampling_rate = PROCESSADOR.feature_extractor.sampling_rate, return_tensors = "pt")

print (inputs)
print (inputs["input_values"])
print (inputs["input_values"].shape)
```

<hr>

```python
with torch.no_grad():
    logits = MODELO (inputs.input_values, attention_mask = inputs.attention_mask).logits

print (logits)
```

<hr>

```python
predicted_logits = torch.argmax (logits, dim = -1)
print (PROCESSADOR.batch_decode (predicted_logits))
```
