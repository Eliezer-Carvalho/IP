```python
from transformers import AutoModelForRNNT, AutoModelForTDT, AutoProcessor

from transformers.audio_utils import load_audio
```

<a href = "https://huggingface.co/nvidia/nemotron-3.5-asr-streaming-0.6b"> <h1> Nvidia Nemotron 3.5 ASR </h1> </a>

```python
MODELO_NAME = "nvidia/nemotron-3.5-asr-streaming-0.6b"

PROCESSADOR = AutoProcessor.from_pretrained (MODELO_NAME)
MODELO = AutoModelForRNNT.from_pretrained (MODELO_NAME)

print (PROCESSADOR)
print (MODELO)
```

<hr>

```python
audio = load_audio (r"C:\Users\Admin\Desktop\whisper\audio_3.wav", sampling_rate = PROCESSADOR.feature_extractor.sampling_rate)

print (audio)
```

<hr>

```python
inputs = PROCESSADOR (audio, sampling_rate = PROCESSADOR.feature_extractor.sampling_rate, language = "pt-PT")

print (inputs)
print (inputs["input_features"].shape)
```

<HR>

```python
output = MODELO.generate (**inputs)

print (output)
```

<hr>

```python
print (PROCESSADOR.decode(output.sequences, skip_special_tokens = True))
```
