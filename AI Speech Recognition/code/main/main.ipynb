```python
from transformers import AutoModelForRNNT, AutoProcessor
from transformers.audio_utils import load_audio


import torch

print (torch.__version__)

device = "cuda" if torch.cuda.is_available() else "cpu"
```

<hr>

```python
MODELO_NAME = "nvidia/nemotron-3.5-asr-streaming-0.6b" #https://huggingface.co/nvidia/nemotron-3.5-asr-streaming-0.6b

try:
    PROCESSADOR = AutoProcessor.from_pretrained (MODELO_NAME) #Tokenizer Wanna Be
    MODELO_SPEECH = AutoModelForRNNT.from_pretrained (MODELO_NAME, device_map = device)

except Exception as e:
    print (e)

## Memory Management

print (torch.cuda.memory_allocated() / 1024**3, "GB")
print (torch.cuda.memory_reserved() / 1024**3, "GB")
```

<hr>

```python
wav = load_audio (r"C:\Users\Admin\Desktop\ip\AI Speech Recognition\audio\audio2.wav", sampling_rate = PROCESSADOR.feature_extractor.sampling_rate)

print (wav)
```

<hr>

```python
inputs = PROCESSADOR (wav, sampling_rate = PROCESSADOR.feature_extractor.sampling_rate, return_tensors = "pt").to(device)

print (inputs)
print (inputs["input_features"].shape)
```

<hr>

```python
outputs = MODELO_SPEECH.generate (**inputs)

print (outputs)
print (outputs["sequences"].shape)
```

<hr>

```python
print (PROCESSADOR.decode (outputs.sequences, skip_special_tokens = True))
```
