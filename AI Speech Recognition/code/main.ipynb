```python
from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC #https://huggingface.co/docs/transformers/v5.14.0/en/model_doc/whisper#whisper

import torch

import librosa
```

```python
MODELO_NOME = "jonatasgrosman/wav2vec2-large-xlsr-53-portuguese"

PROCESSADOR = Wav2Vec2Processor.from_pretrained (MODELO_NOME)
MODELO = Wav2Vec2ForCTC.from_pretrained (MODELO_NOME)

'''
print(MODELO.generation_config.max_length)
print(MODELO.generation_config.max_new_tokens)
print(MODELO.generation_config.eos_token_id)
'''
```

```python
audio, sr = librosa.load (r"C:\Users\Admin\Desktop\whisper\audio_2.wav", sr = 16000)

print (audio)
print (audio.shape) 
```

```python
inputs = PROCESSADOR (audio, sampling_rate = 16000, return_tensors = "pt")

print (inputs.keys())
```

```python
predicted_ids = MODELO(inputs["input_values"])

print (predicted_ids.logits)
print (predicted_ids.logits.shape)
```

```python
ids = torch.argmax(predicted_ids.logits, dim = -1)
transcription = PROCESSADOR.batch_decode(ids)
```
