```python
from transformers.audio_utils import load_audio

from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor
```

<a href = "https://huggingface.co/inesc-id/WhisperLv3-FT-EP-CPP"> <h1> WhisperLv3-FT-EP-CPP </h1> </a> 

```python
MODEL_ID = "inesc-id/WhisperLv3-FT-EP-CPP" # Modelo Encoder -> Decoder Fine Tuning com o dataset CAMÕES
PROCESSOR_ID = "openai/whisper-large-v3" # Processador do modelo original

PROCESSOR = AutoProcessor.from_pretrained (PROCESSOR_ID)
MODELO = AutoModelForSpeechSeq2Seq.from_pretrained (MODEL_ID)
```

<hr>

```python
audio = load_audio (r"C:\Users\Admin\Desktop\ip\AI Speech Recognition\audio\audio2.wav", sampling_rate = PROCESSOR.feature_extractor.sampling_rate)

print (audio)
```

<hr>

```python
inputs = PROCESSOR (audio, sampling_rate = PROCESSOR.feature_extractor.sampling_rate, return_tensors = "pt")

print (inputs)
print (inputs.keys())
```

<hr>

```python
prev_ids = MODELO.generate (**inputs)

print (prev_ids)
```

<hr>

```python
print (PROCESSOR.batch_decode (prev_ids, skip_special_tokens = True))
```
