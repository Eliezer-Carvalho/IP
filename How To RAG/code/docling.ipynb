```python
from docling.chunking import HierarchicalChunker

# 1. Converter o documento
converter = DocumentConverter()

result = converter.convert(r"C:\Users\Admin\Desktop\ip\How To RAG\docs\ABC_Máquinas_Elétricas.pdf")

# 2. Obter o documento estruturado
doc = result.document

# 3. Fazer chunking preservando a hierarquia
chunker = HierarchicalChunker()

chunks = list(chunker.chunk(doc))

# 4. Inspecionar os chunks
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i} ---")
    print(chunk.text)
    print("Metadata:", chunk.meta)
```
