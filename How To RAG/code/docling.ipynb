```python
from docling.document_converter import DocumentConverter
from docling.chunking import HierarchicalChunker

from pathlib import Path

from langchain_core.documents import Document
```

```python
DOCLING = DocumentConverter ()
PATH = r"C:\Users\Admin\Desktop\ip\How To RAG\docs"
PATH = Path (PATH).glob ("*.pdf")

Chunk_Hier = HierarchicalChunker ()

DOCS = []
for documento in PATH:

    CONVERSOR = DOCLING.convert (documento)
    PDF = CONVERSOR.document

    CHUNKS = list(Chunk_Hier.chunk (PDF))

    for chunk in CHUNKS:

        content = Chunk_Hier.contextualize (chunk)
        heading = chunk.meta.headings

        """
        Única maneira de obter o número da página.
        """
        if "Figura" in content:
            continue

        if "<!-- formula-not-decoded -->" in content:
            continue

        if "Índice" in heading:
            continue

        pages = set ()

        for item in chunk.meta.doc_items:
            for prov in item.prov:
                pages.add (prov.page_no)


        DOCS.append (
            Document (
                page_content = content,
                metadata = {
                    "headings": heading,
                    "page_number": sorted (pages)
                }
            )
        )
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

Dense_Retrieval = HuggingFaceEmbeddings (model_name = r"C:\Users\Admin\Desktop\models\Encoders Models\Encoder Models\Qwen 0.6B")
```

```python
from langchain_qdrant import FastEmbedSparse

Sparse_Retrieval = FastEmbedSparse (model_name = "Qdrant/bm25")
```

```python
from langchain_qdrant import QdrantVectorStore, RetrievalMode

Vector_DataBase = QdrantVectorStore.from_documents (
        documents = DOCS, 
        embedding = Dense_Retrieval, 
        sparse_embedding = Sparse_Retrieval, 
        path = "./qdrant_data",
        collection_name = "RAG", 
        retrieval_mode = RetrievalMode.HYBRID
    )
```

```python
query = "Como é gerado o campo magnético ?"

docs = Vector_DataBase.similarity_search (query, k = 50)

display (docs)
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

path = r"C:\Users\Admin\Desktop\models\Encoders Models\Cross Encoder Models\BAAI BGE Reranker V2 M3"

cross_encoder_model = AutoModelForSequenceClassification.from_pretrained (path, device_map = "cuda", dtype = torch.float16)
cross_encoder_model_tokenization = AutoTokenizer.from_pretrained (path)
```

```python
pairs = [[query, doc.page_content] for doc in docs]

#print (pares)
device = "cuda" if torch.cuda.is_available () else "cpu"
inputs = cross_encoder_model_tokenization (pairs, return_tensors = "pt", padding = True, truncation = True).to (device)
#print (inputs)

cross_encoder_model.eval()
with torch.no_grad():
    logits = cross_encoder_model (**inputs).logits
    #print (logits)

#Organiza os logits com os chunks correspondentes #Muita Atenção! Para calcular HitRate e MRR com as funções criadas é necessário estar no mesmo formato do output do RRF.
rerank = sorted (zip (docs, logits.tolist()), key = lambda x: x[1], reverse = True) #x[3] define que arg é usado pelo reverse
rerank = rerank [:5]
```
