```python
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor
from silero_vad import load_silero_vad, get_speech_timestamps
import librosa
import torch

from transformers.audio_utils import load_audio
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
MODEL_PATH = r"C:\Users\Admin\Desktop\models\ASR Models\Whisper\WhisperLv3-PT-All 4Bit"

TOKENIZER = AutoProcessor.from_pretrained (MODEL_PATH)
MODELO = AutoModelForSpeechSeq2Seq.from_pretrained (MODEL_PATH, device_map = device, dtype = torch.float16)
```

```python
VAD_MODEL = load_silero_vad ()
```

```python
path = [r"C:\Users\Admin\Desktop\ip\Automatic Speech Recognition\5.wav", r"C:\Users\Admin\Desktop\ip\Automatic Speech Recognition\audio\audio5.wav"]

for x in path:

    WAV, SAMPLE_RATE = librosa.load (x, sr = 16000, mono = True)

    speech_timestamps = get_speech_timestamps (WAV, VAD_MODEL)

    for y in speech_timestamps:

        WAV2 = WAV[y["start"] : y["end"]]

        inputs = TOKENIZER (WAV2, sampling_rate = TOKENIZER.feature_extractor.sampling_rate, return_tensors = "pt", truncation = False)
        
        inputs = inputs["input_features"].to(device, dtype = torch.float16)

        outputs = MODELO.generate (inputs, return_timestamps = True, task = "transcribe", language = "pt", num_beams = 10)

        print (TOKENIZER.decode (outputs))
        



```
