```python
with open (r"C:\Users\Admin\Desktop\ip\Automatic Speech Recognition\v2\Eval\Semantic-Layer\System_Prompt.md", "r", encoding = "utf-8") as f:
    System_Prompt = f.read ()
```

```python
API = OpenAI (base_url = "http://127.0.0.1:8080", api_key = "IP")

MENSAGENS = [
    {"role": "system", "content": System_Prompt},
    {"role": "user", "content": "Cria exatamente 20 exemplos para o Dataset!"}
]

###
class Expected_Output (BaseModel):
    operation: Literal["RETURN", "COUNT", "MEAN", "SUM"]
    columns: Literal["AUDIO_TIME", "TEMPO_PRÉ_PROCESSAMENTO", "TRANSCRIÇÃO", "TEMPO_PROCESSAMENTO_MODELO_ASR", "TEMPO_INFERÊNCIA_MODELO_ASR", "LATÊNCIA", "TOKENS_PER_SECOND_DECODE", "HARDWARE_LLM", "MODELO_LLM", "AUDITORIA_LLM", "NÚMERO_DE_TOKENS_PROCESSADOS", "TEMPO_PREFILL_LLM", "TOKENS_PER_SECOND_PREFILL_LLM", "TEMPO_DECODE_LLM", "TOKENS_PER_SECOND_DECODE_LLM", "LATÊNCIA_LLM"]

class Structured_Output (BaseModel):
    id: int
    prompt: str
    prompt_eng: str
    expected_output: Expected_Output

class Output (BaseModel):
    exemplos: list[Structured_Output] = Field (min_length = 20, max_length = 20)

###

dataset = []

for i in range (25):
    OUTPUT = API.beta.chat.completions.parse (model = "GPTOSS20B", messages = MENSAGENS, response_format = Output)

    DADOS_CRUS = OUTPUT.model_dump ()
    EXEMPLOS = DADOS_CRUS["choices"][0]["message"]["content"] 

    #print (X)
    dataset.append (EXEMPLOS)
    print (len(dataset))
```

```python
print (len(dataset))

"""
Esta salganhada toda porque o output é uma lista e não dá para converter lista para json da maneira correta por isso iteramos a lista com exemplo a exemplo e criamos json para cada exemplo e depois
juntamos todos.
"""
with open ("dataset.json", "w", encoding = "utf-8") as f:
    for exemplos in dataset:

        exemplos = json.loads(exemplos)

        for exemplo in exemplos["exemplos"]:
            f.write(json.dumps (exemplo, ensure_ascii = False, indent = 4) + "\n")
```
