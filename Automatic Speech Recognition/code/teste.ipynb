```python
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor

from transformers.audio_utils import load_audio

import torch

import time
```

```python
device = "cuda" if torch.cuda.is_available () else "cpu"
```

```python
MODEL = r"C:\Users\Admin\Desktop\models\ASR Models\Whisper\WhisperLv3-PT-All 8Bit"

PROCESSOR_ASR = AutoProcessor.from_pretrained (MODEL)
MODEL_ASR = AutoModelForSpeechSeq2Seq.from_pretrained (MODEL, device_map = device, dtype = torch.float16)
```

```python
WAV = load_audio (r"C:\Users\Admin\Desktop\ip\Automatic Speech Recognition\audio\audio4.wav", sampling_rate = PROCESSOR_ASR.feature_extractor.sampling_rate)
```

```python
inputs = PROCESSOR_ASR (WAV, sampling_rate = PROCESSOR_ASR.feature_extractor.sampling_rate, return_tensors = "pt", truncation = False).to(device) #TRUNCATION OBRIGATÓRIO PARA ÁUDIOS > 30s

print (inputs)
print (inputs["input_features"])
print (inputs["input_features"].shape)
```

```python
ids = MODEL_ASR.generate (inputs["input_features"], return_timestamps = True, task = "transcribe", language = "pt", return_segments = True)
```
