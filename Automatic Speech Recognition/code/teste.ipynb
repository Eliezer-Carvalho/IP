```python
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor

from transformers.audio_utils import load_audio

import torch

import time
```

```python
device = "cuda" if torch.cuda.is_available () else "cpu"
```

```python
MODEL = r"C:\Users\Admin\Desktop\models\ASR Models\Whisper\WhisperLv3-PT-All 4Bit"

PROCESSOR_ASR = AutoProcessor.from_pretrained (MODEL)
MODEL_ASR = AutoModelForSpeechSeq2Seq.from_pretrained (MODEL, device_map = device, dtype = torch.float16)
```

```python
print (MODEL_ASR.dtype)
```

```python
WAV = load_audio (r"C:\Users\Admin\Desktop\ip\Automatic Speech Recognition\audio\audio5.wav", sampling_rate = PROCESSOR_ASR.feature_extractor.sampling_rate)
```

```python
inputs = PROCESSOR_ASR (WAV, sampling_rate = PROCESSOR_ASR.feature_extractor.sampling_rate, return_tensors = "pt", truncation = False) #TRUNCATION OBRIGATÓRIO PARA ÁUDIOS > 30s

inputs = inputs["input_features"].to (device, dtype = torch.float16)

print (inputs)
#print (inputs["input_features"])
#print (inputs["input_features"].shape)
```

```python
ids = MODEL_ASR.generate (inputs, return_timestamps = True, task = "transcribe", language = "pt", num_beams = 8)
```

```python
#print (ids.keys())
#print (ids["sequences"])
#print (ids["segments"])

x = PROCESSOR_ASR.decode (ids)

print (x)
print (len(str(x).split()))




#print (ids["segments"])


"""
for x in ids["segments"][0]:
    print (x["tokens"])
    print (PROCESSOR_ASR.decode (x["tokens"]))
    #print (PROCESSOR_ASR.decode (x["tokens"]))
"""
```
