```python
from transformers import AutoProcessor, AutoModelForSpeechSeq2Seq, GenerationConfig, BitsAndBytesConfig
```

<h1> Update Gen Config </h1>

```python
MODEL = "inesc-id/WhisperLv3-PT-All"

PROCESSOR_ASR = AutoProcessor.from_pretrained (MODEL)
MODEL_ASR = AutoModelForSpeechSeq2Seq.from_pretrained (MODEL)

GEN_CONFIG = GenerationConfig.from_pretrained ("openai/whisper-large-v3")

MODEL_ASR.generation_config = GEN_CONFIG

PROCESSOR_ASR.save_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\teste")
MODEL_ASR.save_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\teste")
```

<h1> Model 8 Bit </h1>

```python
BnB = BitsAndBytesConfig (
    load_in_8bit = True,
)

path = r"C:\Users\Admin\Desktop\models\ASR Models\teste"

PROCESSOR_ASR = AutoProcessor.from_pretrained (path)
MODEL_ASR = AutoModelForSpeechSeq2Seq.from_pretrained (path, quantization_config = BnB)

PROCESSOR_ASR.save_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\teste8B")
MODEL_ASR.save_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\teste8B")
```

<h1> Model 4 Bit </h1>

```python
BnB = BitsAndBytesConfig (
    load_in_4bit = True,
    bnb_4bit_use_double_quant = True,
    bnb_4bit_compute_dtype = 'float16',
    bnb_4bit_quant_type = "nf4"
)

path = r"C:\Users\Admin\Desktop\models\ASR Models\teste"

PROCESSOR_ASR = AutoProcessor.from_pretrained (path)
MODEL_ASR = AutoModelForSpeechSeq2Seq.from_pretrained (path, quantization_config = BnB)

PROCESSOR_ASR.save_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\teste4B")
MODEL_ASR.save_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\teste4B")
```
