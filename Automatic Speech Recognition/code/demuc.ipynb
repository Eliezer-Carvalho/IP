```python
import librosa 
from demucs.api import Separator
import torchaudio 
import torch

from transformers import AutoProcessor, AutoModelForSpeechSeq2Seq
import soundfile as sf
```

```python
SEPARADOR = Separator (model = "htdemucs")
```

```python
WAV, SR = librosa.load (r"C:\Users\Admin\Desktop\ip\Automatic Speech Recognition\audio\audio8.wav", sr = 44100, mono = False) #44100 porque é o que o Demucs precisa de receber

display (WAV)
print (WAV.shape) 
print (type (WAV)) #<class 'numpy.ndarray'>

WAV = torch.from_numpy (WAV)

display (WAV)
print (WAV.shape) 
print (type (WAV))
print (WAV.dtype)

ORIGINAL, SEPARADO = SEPARADOR.separate_tensor (WAV, sr = SR)

print (SEPARADO.keys())

VOZES = SEPARADO ["vocals"]

VOZES = torchaudio.functional.resample (VOZES, orig_freq = SR, new_freq = 16000)
VOZES = VOZES.mean (dim = 0)

display (VOZES)
print (VOZES.shape) 
print (type (VOZES))
print (VOZES.dtype)

sf.write ("x.wav", VOZES, samplerate = 16000)
```

```python
device = "cuda" if torch.cuda.is_available () else "cpu"

PROCESSOR = AutoProcessor.from_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\Whisper\WhisperLv3-PT-All 4Bit")
MODEL_ASR = AutoModelForSpeechSeq2Seq.from_pretrained (r"C:\Users\Admin\Desktop\models\ASR Models\Whisper\WhisperLv3-PT-All 4Bit", device_map = device, dtype = torch.float16)
```

```python
inputs = PROCESSOR (VOZES, sampling_rate = PROCESSOR.feature_extractor.sampling_rate, return_tensors = "pt", truncation = False) # Truncation obrigatório para áudios >30s
inputs = inputs["input_features"].to (device, dtype = torch.float16) # Passar para GPU

with torch.inference_mode ():
    outputs = MODEL_ASR.generate (inputs, return_timestamps = True, task = "transcribe", language = "pt", num_beams = 5) # Beam Search # return_timestamps obrigatório para áudios >30s 

trans = PROCESSOR.batch_decode (outputs, skip_special_tokens = True)[0]
```
